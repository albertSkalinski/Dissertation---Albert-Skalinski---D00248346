## Importing the Libraries

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
```

## Loading Datasets

```python
# Setting paths for the data folder and the orignal data files
dataFolder = "C:/Users/sKALa/Repos/Semester 2/Dissertation/Albert Skalinski - Dissertation/data"
originalData1Path = dataFolder + "/originalData1.csv"
originalData2Path = dataFolder + "/originalData2.csv"

# Reading both .csv files
df1 = pd.read_csv(originalData1Path)
df2 = pd.read_csv(originalData2Path)

# Checking if the data has been read correctly
print("Dataset 1 shape:", df1.shape)
print("Dataset 2 shape:", df2.shape)
```

```python
# Displaying the first few rows of each dataset
display(df1.head())
display(df2.head())
```

## Matching Columns

```python
# The second dataframe has its columns in a different order, so they need to be reordered
df2 = df2[df1.columns]

# Checking applied changes
display(df1.head())
display(df2.head())
```

# Combining Datasets

```python
# Combining both datasets into one
df = pd.concat([df1, df2], ignore_index = True)

# Checking for changes
display(df.shape)
display(df.head())
```

## Resolving Missing Values

```python
# Checking for missing values
missingValues = df.isna().sum()
display(missingValues)
```

```python
# Removing rows with missing values, since they're marginal compared to the size of the dataset
df.dropna(inplace = True)
```

## Resolving Duplicates

```python
# Checking for exact duplicates
exactDuplicates = df.duplicated().sum()

print("Number of exact duplicate rows:", exactDuplicates)
```

```python
# Removing the duplicates
df = df.drop_duplicates()

# Checking for applied changes
print(df.shape)
```

```python
# Checking for duplicate track IDs
duplicateTrackIDs = df["track_id"].duplicated().sum()

print("Number of duplicate track IDs:", duplicateTrackIDs)
```

```python
# Removing duplicate track IDs, keeping only the first occurrence

df = df.drop_duplicates(subset = "track_id", keep = "first")
df = df.reset_index(drop = True)

print("Shape after removing duplicate track IDs:", df.shape)
print("Remaining duplicate IDs:", df["track_id"].duplicated().sum())
```

## Checking Datatypes

```python
# Checking datatypes
print(df.dtypes)
```

```python
# Changing some of the types into ints to better match the data they represent
df["time_signature"] = df["time_signature"].astype(int)
df["track_popularity"] = df["track_popularity"].astype(int)
df["mode"] = df["mode"].astype(int)
df["key"] = df["key"].astype(int)
df["duration_ms"] = df["duration_ms"].astype(int)

# Checking applied changes
print(df.dtypes)
```

```python
# Making duration cleaner by converting it from milliseconds to minutes to two decimal places
df["duration_min"] = df["duration_ms"] / 60000
df["duration_min"] = df["duration_min"].round(2)
```

## Handling Release Date

```python
# Parsing release dates
df["track_album_release_date_original"] = df["track_album_release_date"]
df["track_album_release_date"] = pd.to_datetime(df["track_album_release_date"], errors = "coerce")

# The print statement below indicates all the fields that have a year only instead of a full date
print("Missing parsed release dates:", df["track_album_release_date"].isna().sum())
```

```python
# Adding additional columns for release year and month for further analysis
df["release_year"] = df["track_album_release_date"].dt.year
df["release_month"] = df["track_album_release_date"].dt.month
```

```python
# Removing rows with invalid release years
df = df.dropna(subset = ["release_year"])
```

```python
# Casting the above columns to int after dropping NaN values
df["release_year"] = df["release_year"].astype(int)
df["release_month"] = df["release_month"].astype(int)
```

```python
# Creating a time period column for further analysis
def assignTimePeriod(year):
    if year < 1980:
        return "pre-1980"
    elif year < 1990:
        return "1980s"
    elif year < 2000:
        return "1990s"
    elif year < 2010:
        return "2000s"
    elif year < 2020:
        return "2010s"
    else:
        return "2020s"

df["time_period"] = df["release_year"].apply(assignTimePeriod)
```

## Removing Unnecessary Columns

```python
# Removing columns that were found unncessary for the analysis and provided no useful information
df = df.drop(columns = ["track_href", "uri", "playlist_name", "analysis_url", "track_album_id", "id", "playlist_subgenre", "type",
            "playlist_id", "track_album_name", "track_album_release_date_original", "release_decade", "track_id", "duration_ms"], errors = "ignore")

# Checking for applied changes
display(df.head(10))
```

## Reordering Columns

```python
# Reordering columns for clarity purposes
orderedColumns = [
    "track_name",
    "track_artist",
    "playlist_genre",
    "track_popularity",
    "track_album_release_date",
    "release_year",
    "release_month",
    "time_period",

    "energy",
    "danceability",
    "loudness",
    "liveness",
    "valence",
    "speechiness",
    "instrumentalness",
    "acousticness",

    "tempo",
    "time_signature",
    "mode",
    "key",
    "duration_min"]

df = df[orderedColumns]

df.head()
```

## Chaning Column Names

```python
# Changing the names for clarity purposes
df.rename(columns = {
    "track_name": "name",
    "track_artist": "artist",
    # playlist genre simplified to genre since all the songs in the dataset are from playlists and the genre is the same as the playlist genre
    "playlist_genre": "genre",
    "track_popularity": "popularity",
    # album release date simplified to release date since the dataset only contains album tracks and the release date
    #  is the same as the album release date
    "track_album_release_date": "releaseDate",
    "release_year": "releaseYear",
    "release_month": "releaseMonth",
    "time_period": "timePeriod",

    "energy": "energy",
    "danceability": "danceability",
    "loudness": "loudness",
    "liveness": "liveness",
    "valence": "valence",
    "speechiness": "speechiness",
    "instrumentalness": "instrumentalness",
    "acousticness": "acousticness",

    "tempo": "tempo",
    "time_signature": "timeSignature",
    "mode": "mode",
    "key": "key",
    "duration_min": "durationMin"}, inplace = True)

# Checking for applied changes
display(df.head())
```

## Saving the Clean Dataset

```python
# Saving the dataset to .csv
cleanedDataPath = dataFolder + "/cleanData.csv"
df.to_csv(cleanedDataPath, index = False)
```
